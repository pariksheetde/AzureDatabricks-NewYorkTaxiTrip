#### INGEST `TAXI TRIP FROM VOLUME`
- VOLUMN = `/Volumes/nyctaxi/landing/operational/2026/`

```python
from pyspark.sql.functions import input_file_name, col

yellow_taxi_df = (spark.read.format('parquet')
                      .load('/Volumes/nyctaxi/landing/operational/2026/YellowTaxi/*')
                      .withColumn('file_name', col('_metadata.file_path'))
                      .withColumn('load_time_stamp', current_timestamp())
                )
display(yellow_taxi_df.limit(10))
```
