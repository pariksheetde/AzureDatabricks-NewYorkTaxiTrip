#### VALIDATE `BRONZE` LAYER

```sql
SELECT
COUNT(*) AS cnt
FROM
NYCTAXI.BRONZE.GREEN_TAXI;
```

```sql
SELECT
COUNT(*) AS cnt,
file_name
FROM
NYCTAXI.BRONZE.YELLOW_TAXI
GROUP BY file_name
ORDER BY file_name
LIMIT 20;
```

```sql
-- VALIDATE `NYCTAXI.BRONZE.TAXI_ZONE_LOOKUP`
SELECT
*
FROM
NYCTAXI.BRONZE.TAXI_ZONE_LOOKUP
LIMIT 10;
```

```sql
-- VALIDATE NYCTAXI.AUDIT.PIPELINE_EXECUTION_LOG
SELECT
*
FROM
NYCTAXI.AUDIT.PIPELINE_EXECUTION_LOG
WHERE layer = 'BRONZE';
```
