#### VALIDATE `BRONZE` LAYER
- `NYCTAXI.BRONZE.GREEN_TAXI`

```sql
SELECT
COUNT(*) AS cnt
FROM
NYCTAXI.BRONZE.GREEN_TAXI;
```

```sql
-- VALIDATE ALL THE REQUIRED COLUMNS ARE LOADED
SELECT
*
FROM
NYCTAXI.BRONZE.GREEN_TAXI
LIMIT 1
```

```sql
SELECT
COUNT(*) AS cnt,
file_name
FROM
NYCTAXI.BRONZE.GREEN_TAXI
GROUP BY file_name
ORDER BY file_name
LIMIT 20;
```
