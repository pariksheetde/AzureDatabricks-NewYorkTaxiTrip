### VALIDATE `BRONZE` LAYER

#### VALIDATE `NYCTAXI.BRONZE.YELLOW_TAXI`

```sql
SELECT
COUNT(*) AS cnt
FROM
NYCTAXI.BRONZE.YELLOW_TAXI;
```

```sql
SELECT
COUNT(*) AS cnt,
file_name
FROM
NYCTAXI.BRONZE.YELLOW_TAXI
GROUP BY file_name
ORDER BY file_name
LIMIT 20;
```

#### VALIDATE `NYCTAXI.BRONZE.TAXI_ZONE_LOOKUP`

```sql
SELECT
*
FROM
NYCTAXI.BRONZE.TAXI_ZONE_LOOKUP
LIMIT 10;
```

#### VALIDATE `NYCTAXI.SILVER.YELLOW_TAXI_TRIPS_CLEANSED`

```sql
SELECT
*
FROM
NYCTAXI.SILVER.YELLOW_TAXI_TRIPS_CLEANSED
LIMIT 5;
```

#### VALIDATE `NYCTAXI.SILVER.YELLOW_TAXI_TRIPS_ENRICHED`

#### VALIDATE SLOWLY CHANGING DIMENSION
- NYCTAXI.SILVER.REGIONS_SCD_2

```sql
SELECT 
* 
FROM
NYCTAXI.SILVER.REGIONS_SCD_2 
ORDER BY ID ASC, effective_end_date DESC;
```
